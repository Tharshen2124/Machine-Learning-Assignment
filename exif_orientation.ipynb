```python
from PIL import Image, ExifTags
import matplotlib.pyplot as plt

# Use the filename visible in your explorer sidebar
image_path = "test.jpg"

def check_local_orientation(image_path):
    img = Image.open(image_path)

    # Display image
    plt.imshow(img)
    plt.axis("off")
    plt.title(image_path)
    plt.show()

    print(img.mode)

    # Extract EXIF data
    exif_data = img._getexif()

    if exif_data is None:
        print("No EXIF data found.")
    else:
        print("EXIF Data:\n")
        for tag_id, value in exif_data.items():
            tag = ExifTags.TAGS.get(tag_id, tag_id)
            print(f"{tag}: {value}")

check_local_orientation(image_path)
```
